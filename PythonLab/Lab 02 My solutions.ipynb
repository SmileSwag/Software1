```python
# Your code starts here
x=np.arange(0 , 10, 0.001)
y1=np.sqrt( 3.4 * x )
y2=( x - 2.84 ) **2

plt.figure()
plt.plot(x, y1, label='f(x) = sqrt(3.4 * x)')
plt.plot(x, y2, label='g(x) = (x - 2.84)^2')
plt.title("Overview: Both functions visible")
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.scatter([1.367,4.858], [2.17,4.10], color='red')
plt.legend()
plt.show()

fix,ax=plt.subplots(2,figsize=(6,8))

plt.sca(ax[0])
plt.plot(x, y1, label='f(x) = sqrt(3.4 * x)')
plt.plot(x, y2, label='g(x) = (x - 2.84)^2')
plt.xlim(1.2,1.5) 
plt.ylim(0, 5) 
plt.title("Intersection Point 1 (zoomed)")
plt.scatter(1.367, 2.17, color='red')
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.legend()

plt.sca(ax[1])
plt.plot(x, y1, label='f(x) = sqrt(3.4 * x)')
plt.plot(x, y2, label='g(x) = (x - 2.84)^2')
plt.xlim(4.7,5.0) 
plt.ylim(0, 8) 
plt.title("Intersection Point 2 (zoomed)")
plt.scatter(4.856, 4.1, color='red')
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.legend()
plt.show()



```
