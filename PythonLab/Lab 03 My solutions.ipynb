### My solution

```python
# start your code here
x = np.linspace(-10, 10, 2000)
y = (((-0.7 <= x) & (x < 3.3)) * (0.7 + x) + 
     (((3.3 <= x) & (x < 7.3)) * (7.3 - x)))

plt.figure(figsize=(8, 5))
plt.plot(x, y, label='f(x)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Problem 1: Piecewise Function')
plt.legend()
plt.grid(True)
plt.show()
```
